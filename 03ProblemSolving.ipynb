```
totalNumbers = input()
array = input().split()

lowestNumber = int(array[0])
lowestIndex = 0

for i in range(int(totalNumbers)):
  if int(array[i]) < lowestNumber:
    lowestNumber = int(array[i])
    lowestIndex = i + 1

if lowestIndex == 0:
  lowestIndex = 1

print(f"{lowestNumber} {lowestIndex}")
```

```
numberOfElements = input()
array = input().split()
array = list(map(int, array))

maximumValue = int(array[0])
maximumValueIndex = 0
minValue = int(array[0])
minValueIndex = 0

for i in range(int(numberOfElements)):
  if array[i] > maximumValue:
    maximumValue = array[i]
    maximumValueIndex = i

  if array[i] < minValue:
    minValue = array[i]
    minValueIndex = i

array[maximumValueIndex] = minValue
array[minValueIndex] = maximumValue

for i in array:
  print(i, end = " ")
```
