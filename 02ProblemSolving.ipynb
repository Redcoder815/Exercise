```
length = input().split()
inputArray = input().split()
inputSetA = set(input().split())
inputSetB = set(input().split())

happiness = 0

for i in inputArray:
  if i in  inputSetA:
    happiness = happiness + 1

for i in inputArray:
  if i in inputSetB:
    happiness = happiness - 1

print(happiness)
```
