```
numbers = input()
nor = numbers.split()

norm = (float(nor[0]) - float(nor[1])) / (float(nor[2]) - float(nor[1]))
print(f"{norm:.2f}")
```

```
n, target = input(), input()
weightsSum = 0.00

for i in range(int(n)):
  val = float(input())
  weightsSum = weightsSum + float(val)

averageWeightsLoss = weightsSum / int(n)

if averageWeightsLoss <= float(target):
  print("PASS")
else:
  print("RETRY")
```

```
n = input()

yes, no = 0, 0
for i in range(int(n)):
  vote = input()
  if vote.upper() == "YES":
    yes = yes + 1
  elif vote.upper() == "NO":
    no = no + 1

if yes >= no:
  print("ACCEPT")
else:
  print("REJECT")
```

```
emotion = input().lower()

happy = ["happy", "joy", "smile"]
sad = ["sad", "cry", "angry"]

if (any(word in emotion for word in happy)):
  print("Happy Mood")
elif (any(word in emotion for word in sad)):
  print("Sad Mood")
else:
  print("Neutral Mood")
```
