```
words = input()

wordsAsList = list(words)

count = {}

for character in wordsAsList:
  count[character] = count.get(character, 0) + 1

sortedDictionary = dict(sorted(count.items()))

for key, value in sortedDictionary.items():
  print(f"{key} : {value}")
```
