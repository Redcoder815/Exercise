```
numbers = input()
nor = numbers.split()

norm = (float(nor[0]) - float(nor[1])) / (float(nor[2]) - float(nor[1]))
print(f"{norm:.2f}")
```

```
n, target = input(), input()
weightsSum = 0.00

for i in range(int(n)):
  val = float(input())
  weightsSum = weightsSum + float(val)

averageWeightsLoss = weightsSum / int(n)

if averageWeightsLoss <= float(target):
  print("PASS")
else:
  print("RETRY")
```

```
n = input()

yes, no = 0, 0
for i in range(int(n)):
  vote = input()
  if vote.upper() == "YES":
    yes = yes + 1
  elif vote.upper() == "NO":
    no = no + 1

if yes >= no:
  print("ACCEPT")
else:
  print("REJECT")
```

```
emotion = input().lower()

happy = ["happy", "joy", "smile"]
sad = ["sad", "cry", "angry"]

if (any(word in emotion for word in happy)):
  print("Happy Mood")
elif (any(word in emotion for word in sad)):
  print("Sad Mood")
else:
  print("Neutral Mood")
```

```
brightness = input()

averageBrightness = 0.0

brightnessArray = brightness.split()

for i in  range(len(brightnessArray)):
  averageBrightness = averageBrightness + float(brightnessArray[i])

avgBri = averageBrightness / len(brightnessArray)

if avgBri < 85:
  print("Dark Image")
elif avgBri>= 85 and avgBri <= 170:
  print("Normal Image")
else:
  print("Bright Image")
```

```
prediction = input()

predictionList = prediction.split()

howManyA = 0
howManyB = 0

for i in range(len(predictionList)):
  if predictionList[i] == "A":
    howManyA = howManyA + 1
  else:
    howManyB = howManyB + 1

percentageA = (howManyA / len(predictionList)) * 100
percentageB = (howManyB / len(predictionList)) * 100

if percentageA > 70 or percentageB > 70:
  print("Biased Model")
else:
  print("Fair Model")
```

```
sentence = input()
split_sentence = sentence.split()

data = ['ai', 'data', 'model', 'learn', 'train', 'neural']

match_count = len([word for word in data if word in split_sentence])

if match_count >= 2:
  print("AI Detected")
else:
  print("Not AI Related")
```
